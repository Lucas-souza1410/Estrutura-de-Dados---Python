```python
def classificar_questao(nota):
    # Verifica se a nota está fora do intervalo permitido
    if nota < 0 or nota > 10:
        return "Nota inválida. Digite um valor entre 0 e 10."
    # Classificação por faixa de nota
    elif nota >= 9.0:
        return "Questão A"
    elif nota >= 8.0:
        return "Questão B"
    elif nota >= 7.0:
        return "Questão C"
    elif nota >= 6.0:
        return "Questão D"
    else:
        return "Questão F"

# Entrada do usuário com tratamento de erro
try:
    nota = float(input("Digite a nota da questão (0 a 10): "))
    resultado = classificar_questao(nota)
    print(resultado)
except ValueError:
    print("Por favor, digite um número válido.")

```

Aula 05 - Alocação dinâmica:

```python
class TextoDinamico:
    def __init__(self):
        self.paragrafos = []

    def adicionar_paragrafo(self, texto):
        self.paragrafos.append(texto)

    def remover_paragrafo(self, indice):
        if 0 <= indice < len(self.paragrafos):
            self.paragrafos.pop(indice)

    def editar_paragrafo(self, indice, novo_texto):
        if 0 <= indice < len(self.paragrafos):
            self.paragrafos[indice] = novo_texto

    def mostrar_documento(self):
        for i, p in enumerate(self.paragrafos):
            print(f"{i}: {p}")
```

```python
class MatrizDinamica:
    def __init__(self, linhas, colunas):
        self.matriz = [[0 for _ in range(colunas)] for _ in range(linhas)]

    def set_valor(self, linha, coluna, valor):
        self.matriz[linha][coluna] = valor

    def get_valor(self, linha, coluna):
        return self.matriz[linha][coluna]

    def mostrar_matriz(self):
        for linha in self.matriz:
            print(linha)

    def somar(self, outra):
        if len(self.matriz) != len(outra.matriz) or len(self.matriz[0]) != len(outra.matriz[0]):
            raise ValueError("Dimensões incompatíveis.")
        resultado = MatrizDinamica(len(self.matriz), len(self.matriz[0]))
        for i in range(len(self.matriz)):
            for j in range(len(self.matriz[0])):
                valor = self.matriz[i][j] + outra.matriz[i][j]
                resultado.set_valor(i, j, valor)
        return resultado
```

```python
class BibliotecaDinamica:
    def __init__(self):
        self.livros = []

    def adicionar_livro(self, titulo):
        self.livros.append(titulo)

    def remover_livro(self, titulo):
        if titulo in self.livros:
            self.livros.remove(titulo)

    def buscar_livro(self, titulo):
        return titulo in self.livros

    def listar_livros(self):
        for livro in self.livros:
            print(livro)
```

Aula 06 - Lista Linear:

```python
class No:
    def __init__(self, valor):
        self.valor = valor
        self.proximo = None

class ListaEncadeada:
    def __init__(self):
        self.cabeca = None

    def inserir_inicio(self, valor):
        novo_no = No(valor)
        novo_no.proximo = self.cabeca
        self.cabeca = novo_no

    def exibir(self):
        atual = self.cabeca
        while atual:
            print(atual.valor, end=" -> ")
            atual = atual.proximo
        print("None")

    def inverter(self):
        anterior = None
        atual = self.cabeca
        while atual:
            proximo = atual.proximo
            atual.proximo = anterior
            anterior = atual
            atual = proximo
        self.cabeca = anterior
```

```python
class ListaEncadeada:
    # ... (demais métodos como __init__, inserir_inicio, exibir, inverter)

    def remover_duplicatas(self):
        valores_vistos = set()
        atual = self.cabeca
        anterior = None

        while atual:
            if atual.valor in valores_vistos:
                # Duplicado: remover o nó atual
                anterior.proximo = atual.proximo
            else:
                # Primeiro valor encontrado: adicionar ao set
                valores_vistos.add(atual.valor)
                anterior = atual
            atual = atual.proximo
```

Aula 09 - Pilhas:

```python
def inverter_string_com_pilha(s):
    pilha = []

    # Empilhar todos os caracteres da string
    for char in s:
        pilha.append(char)

    # Desempilhar e construir a string invertida
    string_invertida = ""
    while pilha:
        string_invertida += pilha.pop()

    return string_invertida
```

Aula 10 - Filas:

```python
from collections import deque

def inverter_fila(fila):
    pilha = []

    # Desenfileira os elementos da fila e empilha na pilha
    while fila:
        pilha.append(fila.popleft())

    # Desempilha da pilha e reenfileira na fila
    while pilha:
        fila.append(pilha.pop())
```

```python
class FilaComDoisPilhas:
    def __init__(self):
        self.pilha1 = []  # para enfileirar
        self.pilha2 = []  # para desenfileirar

    def enfileirar(self, item):
        self.pilha1.append(item)

    def desenfileirar(self):
        if self.esta_vazia():
            raise IndexError("A fila está vazia.")

        if not self.pilha2:
            while self.pilha1:
                self.pilha2.append(self.pilha1.pop())

        return self.pilha2.pop()

    def esta_vazia(self):
        return not self.pilha1 and not self.pilha2
```
